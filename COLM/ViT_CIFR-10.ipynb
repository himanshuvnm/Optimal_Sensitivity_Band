# Tiny ViT model for CIFAR-10

```python
import os, math, time, json, random
from pathlib import Path

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision
import torchvision.transforms as T

import matplotlib.pyplot as plt
from tqdm.auto import tqdm

# ---- Reproducibility ----
def seed_all(seed=0):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

seed_all(0)

# ---- Device ----
if torch.backends.mps.is_available():
    device = torch.device("mps")
elif torch.cuda.is_available():
    device = torch.device("cuda")
else:
    device = torch.device("cpu")
print("device:", device)

# ---- High-quality plot defaults ----
plt.rcParams.update({
    "figure.dpi": 200,
    "savefig.dpi": 300,
    "font.size": 11,
    "axes.titlesize": 12,
    "axes.labelsize": 11,
    "legend.fontsize": 10,
    "lines.linewidth": 2.0,
})
```

```python
DATA_ROOT = Path("./data").resolve()
DATA_ROOT.mkdir(parents=True, exist_ok=True)

# CIFAR-10 normalization (standard)
CIFAR_MEAN = (0.4914, 0.4822, 0.4465)
CIFAR_STD  = (0.2470, 0.2435, 0.2616)

train_tf = T.Compose([
    T.RandomCrop(32, padding=4),
    T.RandomHorizontalFlip(),
    T.ToTensor(),
    T.Normalize(CIFAR_MEAN, CIFAR_STD),
])

test_tf = T.Compose([
    T.ToTensor(),
    T.Normalize(CIFAR_MEAN, CIFAR_STD),
])

train_ds = torchvision.datasets.CIFAR10(root=str(DATA_ROOT), train=True, download=True, transform=train_tf)
test_ds  = torchvision.datasets.CIFAR10(root=str(DATA_ROOT), train=False, download=True, transform=test_tf)

batch_size = 128
num_workers = 0  # mac safe; increase if stable on your machine

train_loader = torch.utils.data.DataLoader(train_ds, batch_size=batch_size, shuffle=True, num_workers=num_workers, pin_memory=False)
test_loader  = torch.utils.data.DataLoader(test_ds,  batch_size=256, shuffle=False, num_workers=num_workers, pin_memory=False)

print(len(train_ds), len(test_ds))
```

```python
import torch
import torch.nn as nn
import torchvision

def make_tiny_vit_cifar(num_classes=10):
    """
    Small Vision Transformer for 32x32 CIFAR-10.
    """
    model = torchvision.models.vision_transformer.VisionTransformer(
        image_size=32,
        patch_size=4,
        num_layers=6,
        num_heads=8,
        hidden_dim=256,
        mlp_dim=512,
        num_classes=num_classes,
        dropout=0.1,
        attention_dropout=0.1,
        representation_size=None,
        norm_layer=lambda dim: nn.LayerNorm(dim, eps=1e-6),
    )
    return model

vit = make_tiny_vit_cifar().to(device)
print("Tiny ViT params:", sum(p.numel() for p in vit.parameters() if p.requires_grad))
```

```python
# Clean accuracy

@torch.no_grad()
def eval_acc(model, loader):
    model.eval()
    correct, total = 0, 0
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        logits = model(x)
        pred = logits.argmax(dim=1)
        correct += (pred == y).sum().item()
        total += y.numel()
    return correct / total
```

```python
@torch.no_grad()
def _unit_noise_like(x):
    v = torch.randn_like(x)
    B = x.shape[0]
    denom = v.flatten(1).norm(dim=1, keepdim=True).view(B, *([1]*(x.ndim-1))) + 1e-12
    return v / denom

def sensitivity_sigma_mean_fd(model, x, eps=1e-3, n_vecs=2):
    """
    Returns scalar sigma_mean proxy on batch x.
    Uses forward passes only (no higher-order derivatives).
    """
    model.eval()
    x = x.detach()
    with torch.no_grad():
        fx = model(x)  # (B,K)
    vals = []
    for _ in range(n_vecs):
        v = _unit_noise_like(x)
        with torch.no_grad():
            fxp = model(x + eps*v)
        Jv = (fxp - fx) / eps  # (B,K)
        # L2 norm across classes
        vals.append(Jv.norm(dim=1))  # (B,)
    vals = torch.stack(vals, dim=0)  # (n_vecs,B)
    # proxy for operator norm: max over random directions, then mean over batch
    per_ex = vals.max(dim=0).values
    return float(per_ex.mean().item())
```

```python
C10C_CORRUPTIONS = [
    "gaussian_noise","shot_noise","impulse_noise",
    "defocus_blur","glass_blur","motion_blur","zoom_blur",
    "snow","frost","fog","brightness",
    "contrast","elastic_transform","pixelate","jpeg_compression",
]

def load_c10c_corruption(c10c_dir, corruption, severity):
    """
    Returns (images, labels) for a corruption at a given severity.
    severity in {1..5}. Each severity is a slice of 10k images.
    """
    c10c_dir = Path(c10c_dir)
    x = np.load(c10c_dir / f"{corruption}.npy")  # (50000,32,32,3), uint8
    y = np.load(c10c_dir / "labels.npy")         # (50000,)
    assert 1 <= severity <= 5
    start = (severity-1)*10000
    end   = severity*10000
    return x[start:end], y[start:end]

class CIFAR10CDataset(torch.utils.data.Dataset):
    def __init__(self, x_uint8, y, transform):
        self.x = x_uint8
        self.y = y
        self.transform = transform

    def __len__(self):
        return len(self.y)

    def __getitem__(self, i):
        img = self.x[i]  # uint8 HWC
        img = torchvision.transforms.functional.to_pil_image(img)
        img = self.transform(img)
        return img, int(self.y[i])

@torch.no_grad()
def eval_c10c(model, c10c_dir, severity=5, corruptions=None, batch_size=256):
    """
    Evaluate mean accuracy over corruptions at a severity.
    Returns mean, std across corruptions.
    """
    if corruptions is None:
        corruptions = C10C_CORRUPTIONS

    accs = []
    model.eval()
    for corr in corruptions:
        x, y = load_c10c_corruption(c10c_dir, corr, severity)
        ds = CIFAR10CDataset(x, y, transform=test_tf)
        loader = torch.utils.data.DataLoader(ds, batch_size=batch_size, shuffle=False, num_workers=0)
        a = eval_acc(model, loader)
        accs.append(a)
    return float(np.mean(accs)), float(np.std(accs)), {c:a for c,a in zip(corruptions, accs)}
```

```python
import requests
import tarfile

C10C_ROOT = Path("./cifar10c").resolve()
C10C_ROOT.mkdir(parents=True, exist_ok=True)

CIFAR10C_URL = "https://zenodo.org/record/2535967/files/CIFAR-10-C.tar?download=1"

def download_cifar10c(root=C10C_ROOT):
    root = Path(root)
    tar_path = root / "CIFAR-10-C.tar"
    extracted_dir = root / "CIFAR-10-C"
    if extracted_dir.exists():
        print("CIFAR-10-C already extracted at:", extracted_dir)
        return extracted_dir

    if not tar_path.exists():
        print("Downloading CIFAR-10-C tar...")
        r = requests.get(CIFAR10C_URL, stream=True)
        r.raise_for_status()
        total = int(r.headers.get("content-length", 0))
        with open(tar_path, "wb") as f:
            pbar = tqdm(total=total, unit="B", unit_scale=True, desc="download")
            for chunk in r.iter_content(chunk_size=1<<20):
                if chunk:
                    f.write(chunk)
                    pbar.update(len(chunk))
            pbar.close()
        print("Downloaded:", tar_path)

    print("Extracting...")
    with tarfile.open(tar_path, "r") as tf:
        tf.extractall(path=root)
    print("Extracted to:", extracted_dir)
    return extracted_dir

c10c_dir = download_cifar10c(C10C_ROOT)
```

```python
def train_vit(
    tag,
    lr,
    weight_decay,
    steps=2000,
    log_every=250,
    eps_fd=1e-3,
    n_vecs=2,
    sens_batch_cap=64,
):
    model = make_tiny_vit_cifar().to(device)
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    logs = {
        "step": [],
        "train_loss": [],
        "test_acc": [],
        "sigma_mean": [],
        "c10c_acc_mean": [],
        "c10c_acc_std": [],
    }

    it = iter(train_loader)
    pbar = tqdm(range(1, steps+1), desc=f"train {tag}")

    for step in pbar:
        try:
            x, y = next(it)
        except StopIteration:
            it = iter(train_loader)
            x, y = next(it)

        x, y = x.to(device), y.to(device)

        model.train()
        logits = model(x)
        loss = F.cross_entropy(logits, y)

        opt.zero_grad(set_to_none=True)
        loss.backward()
        opt.step()

        if step % log_every == 0:
            tloss = float(loss.item())
            ta = eval_acc(model, test_loader)

            xb, _ = next(iter(test_loader))
            xb = xb.to(device)[:sens_batch_cap]
            sigma = sensitivity_sigma_mean_fd(model, xb, eps=eps_fd, n_vecs=n_vecs)

            c10c_mu, c10c_sd, _ = eval_c10c(model, c10c_dir, severity=5)

            logs["step"].append(step)
            logs["train_loss"].append(tloss)
            logs["test_acc"].append(ta)
            logs["sigma_mean"].append(sigma)
            logs["c10c_acc_mean"].append(c10c_mu)
            logs["c10c_acc_std"].append(c10c_sd)

            pbar.set_postfix({
                "loss": f"{tloss:.3f}",
                "test": f"{ta:.3f}",
                "sigma": f"{sigma:.3f}",
                "c10c5": f"{c10c_mu:.3f}",
            })

    return model, logs
```

```python
vit_stable_model, vit_stable_logs = train_vit(
    tag="vit_stable",
    lr=3e-4,
    weight_decay=5e-2,
    steps=2000,
    log_every=250,
)

vit_unstable_model, vit_unstable_logs = train_vit(
    tag="vit_unstable",
    lr=1e-3,
    weight_decay=0.0,
    steps=2000,
    log_every=250,
)
```

## helper to unnormalize and plot images

```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import torchvision.transforms.functional as TF

CIFAR_MEAN = (0.4914, 0.4822, 0.4465)
CIFAR_STD  = (0.2470, 0.2435, 0.2616)

classes = [
    "airplane", "automobile", "bird", "cat", "deer",
    "dog", "frog", "horse", "ship", "truck"
]

def unnormalize(img_tensor, mean=CIFAR_MEAN, std=CIFAR_STD):
    """
    img_tensor: [3,H,W] normalized
    returns: [H,W,3] numpy in [0,1]
    """
    x = img_tensor.detach().cpu().clone()
    for c in range(3):
        x[c] = x[c] * std[c] + mean[c]
    x = torch.clamp(x, 0.0, 1.0)
    return x.permute(1, 2, 0).numpy()
```

## build a small CIFAR-10-C batch and show predictions

```python
@torch.no_grad()
def visualize_vit_predictions_on_c10c(
    model,
    c10c_dir,
    corruption="gaussian_noise",
    severity=5,
    n=8,
    transform=test_tf,
    figsize=(14, 6),
):
    model.eval()

    x_raw, y = load_c10c_corruption(c10c_dir, corruption, severity)
    idx = np.random.choice(len(y), size=n, replace=False)

    imgs = []
    labels = []
    for i in idx:
        img = torchvision.transforms.functional.to_pil_image(x_raw[i])
        imgs.append(transform(img))
        labels.append(int(y[i]))

    xb = torch.stack(imgs).to(device)
    yb = torch.tensor(labels, device=device)

    logits = model(xb)
    probs = torch.softmax(logits, dim=-1)
    conf, pred = probs.max(dim=-1)

    fig, axes = plt.subplots(2, int(np.ceil(n/2)), figsize=figsize)
    axes = axes.flatten()

    for i, ax in enumerate(axes[:n]):
        ax.imshow(unnormalize(xb[i]))
        true_name = classes[yb[i].item()]
        pred_name = classes[pred[i].item()]
        color = "green" if pred[i] == yb[i] else "red"
        ax.set_title(
            f"true: {true_name}\npred: {pred_name}\nconf: {conf[i].item():.2f}",
            color=color,
            fontsize=10
        )
        ax.axis("off")

    for ax in axes[n:]:
        ax.axis("off")

    plt.suptitle(f"ViT on CIFAR-10-C: {corruption}, severity={severity}", y=1.02)
    plt.tight_layout()
    plt.show()
```

# Visualize which corruptions break the ViT most

```python
@torch.no_grad()
def corruption_breakdown_vit(model, c10c_dir, severity=5, batch_size=256):
    model.eval()
    accs = {}

    for corr in C10C_CORRUPTIONS:
        x_raw, y = load_c10c_corruption(c10c_dir, corr, severity)
        ds = CIFAR10CDataset(x_raw, y, transform=test_tf)
        dl = torch.utils.data.DataLoader(ds, batch_size=batch_size, shuffle=False, num_workers=0)

        correct, total = 0, 0
        for xb, yb in dl:
            xb, yb = xb.to(device), yb.to(device)
            pred = model(xb).argmax(dim=-1)
            correct += (pred == yb).sum().item()
            total += yb.numel()

        accs[corr] = correct / total

    return accs
```

```python
def plot_corruption_breakdown(accs, title="ViT robustness by corruption type"):
    items = sorted(accs.items(), key=lambda kv: kv[1])
    names = [k for k, _ in items]
    vals = [v for _, v in items]

    plt.figure(figsize=(10, 5),dpi=500)
    plt.bar(range(len(vals)), vals)
    plt.xticks(range(len(vals)), names, rotation=45, ha="right")
    plt.ylabel("Accuracy")
    plt.title(title)
    plt.tight_layout()
    plt.show()
```

# Better than raw attention: patch sensitivity map

```python
@torch.no_grad()
def vit_patch_occlusion_map(model, img_tensor, target_class=None, patch=4):
    """
    img_tensor: [3,32,32] normalized
    returns heatmap [H/patch, W/patch]
    """
    model.eval()
    x = img_tensor.unsqueeze(0).to(device)

    logits = model(x)
    if target_class is None:
        target_class = logits.argmax(dim=-1).item()

    base_score = logits[0, target_class].item()

    H, W = img_tensor.shape[1], img_tensor.shape[2]
    gh, gw = H // patch, W // patch
    heat = np.zeros((gh, gw), dtype=np.float32)

    for i in range(gh):
        for j in range(gw):
            x_occ = x.clone()
            h0, h1 = i * patch, (i + 1) * patch
            w0, w1 = j * patch, (j + 1) * patch

            # zero-out patch in normalized space
            x_occ[:, :, h0:h1, w0:w1] = 0.0

            score = model(x_occ)[0, target_class].item()
            heat[i, j] = base_score - score

    return heat
```

```python
import os
from pathlib import Path
from tqdm import tqdm
import torch
import torch.nn.functional as F

def train_vit(
    tag,
    lr,
    weight_decay,
    steps=2000,
    log_every=250,
    eps_fd=1e-3,
    n_vecs=2,
    sens_batch_cap=64,
    save_dir=None,
    save_every=250,
):
    model = make_tiny_vit_cifar().to(device)
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    # Prepare checkpoint directory
    ckpt_dir = None
    if save_dir is not None:
        ckpt_dir = Path(save_dir) / tag
        ckpt_dir.mkdir(parents=True, exist_ok=True)

    logs = {
        "step": [],
        "train_loss": [],
        "test_acc": [],
        "sigma_mean": [],
        "c10c_acc_mean": [],
        "c10c_acc_std": [],
    }

    it = iter(train_loader)
    pbar = tqdm(range(1, steps + 1), desc=f"train {tag}")

    for step in pbar:
        try:
            x, y = next(it)
        except StopIteration:
            it = iter(train_loader)
            x, y = next(it)

        x, y = x.to(device), y.to(device)

        model.train()
        logits = model(x)
        loss = F.cross_entropy(logits, y)

        opt.zero_grad(set_to_none=True)
        loss.backward()
        opt.step()

        # Save checkpoint
        if ckpt_dir is not None and step % save_every == 0:
            torch.save(
                {
                    "step": step,
                    "tag": tag,
                    "model_state_dict": model.state_dict(),
                    "optimizer_state_dict": opt.state_dict(),
                    "lr": lr,
                    "weight_decay": weight_decay,
                },
                ckpt_dir / f"checkpoint_step_{step}.pt",
            )

        # Log metrics
        if step % log_every == 0:
            tloss = float(loss.item())
            ta = eval_acc(model, test_loader)

            xb, _ = next(iter(test_loader))
            xb = xb.to(device)[:sens_batch_cap]
            sigma = sensitivity_sigma_mean_fd(model, xb, eps=eps_fd, n_vecs=n_vecs)

            c10c_mu, c10c_sd, _ = eval_c10c(model, c10c_dir, severity=5)

            logs["step"].append(step)
            logs["train_loss"].append(tloss)
            logs["test_acc"].append(ta)
            logs["sigma_mean"].append(sigma)
            logs["c10c_acc_mean"].append(c10c_mu)
            logs["c10c_acc_std"].append(c10c_sd)

            pbar.set_postfix({
                "loss": f"{tloss:.3f}",
                "test": f"{ta:.3f}",
                "sigma": f"{sigma:.3f}",
                "c10c5": f"{c10c_mu:.3f}",
            })

    # Save final checkpoint too
    if ckpt_dir is not None:
        torch.save(
            {
                "step": steps,
                "tag": tag,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": opt.state_dict(),
                "lr": lr,
                "weight_decay": weight_decay,
                "logs": logs,
            },
            ckpt_dir / "checkpoint_final.pt",
        )

    return model, logs
```

```python
from pathlib import Path
import torch

def load_checkpoint_model(ckpt_path, model_fn, device):
    ckpt = torch.load(ckpt_path, map_location=device)
    model = model_fn().to(device)
    model.load_state_dict(ckpt["model_state_dict"])
    model.eval()
    return model, ckpt["step"]

def sorted_checkpoint_paths(ckpt_dir):
    ckpt_dir = Path(ckpt_dir)
    paths = sorted(
        [p for p in ckpt_dir.glob("checkpoint_step_*.pt")],
        key=lambda p: int(p.stem.split("_")[-1])
    )
    return paths
```

```python
def make_probe_images(dataset, probe_indices):
    probe_images = []
    probe_labels = []
    for idx in probe_indices:
        x, y = dataset[idx]
        probe_images.append(x.unsqueeze(0))  # [1, C, H, W]
        probe_labels.append(y)
    return probe_images, probe_labels
```

```python
import os
import numpy as np
import torch
from torch.utils.data import Dataset
from torchvision import transforms

class CIFAR10CSeverityDataset(Dataset):
    def __init__(self, c10c_dir, corruption="fog", severity=5, transform=None):
        self.c10c_dir = c10c_dir
        self.corruption = corruption
        self.severity = severity
        self.transform = transform

        x_path = os.path.join(c10c_dir, f"{corruption}.npy")
        y_path = os.path.join(c10c_dir, "labels.npy")

        x_all = np.load(x_path)          # shape [50000, 32, 32, 3]
        y_all = np.load(y_path)          # shape [50000]

        # CIFAR-10-C stores 5 severity blocks, each of size 10000
        block = 10000
        start = (severity - 1) * block
        end = severity * block

        self.x = x_all[start:end]
        self.y = y_all[start:end]

        if self.transform is None:
            self.transform = transforms.Compose([
                transforms.ToTensor(),
            ])

    def __len__(self):
        return len(self.x)

    def __getitem__(self, idx):
        img = self.x[idx]   # uint8, HWC
        label = int(self.y[idx])
        img = self.transform(img)
        return img, label
```

## Occlusion map code

```python
import numpy as np
import torch

def occlusion_map_for_image(
    model,
    image_tensor,      # [1, C, H, W]
    patch_size=4,
    stride=4,
    fill_value=0.0,
    device="cuda"
):
    model.eval()
    x = image_tensor.to(device)

    with torch.no_grad():
        logits = model(x)
        pred_class = logits.argmax(dim=1).item()
        base_score = logits[0, pred_class].item()

    _, _, H, W = x.shape
    out_h = (H - patch_size) // stride + 1
    out_w = (W - patch_size) // stride + 1
    sens_map = np.zeros((out_h, out_w), dtype=np.float32)

    for i in range(out_h):
        for j in range(out_w):
            x_occ = x.clone()
            h0 = i * stride
            w0 = j * stride
            x_occ[:, :, h0:h0+patch_size, w0:w0+patch_size] = fill_value

            with torch.no_grad():
                occ_logits = model(x_occ)
                occ_score = occ_logits[0, pred_class].item()

            sens_map[i, j] = base_score - occ_score

    return sens_map

def occlusion_maps_for_probe_set(
    model,
    probe_images,
    patch_size=4,
    stride=4,
    fill_value=0.0,
    device="cuda"
):
    maps = []
    for img in probe_images:
        m = occlusion_map_for_image(
            model,
            img,
            patch_size=patch_size,
            stride=stride,
            fill_value=fill_value,
            device=device
        )
        maps.append(m)
    return np.stack(maps, axis=0)   # [N, H', W']
```

```python
def normalize_map(m):
    m = np.asarray(m, dtype=float)
    m = np.maximum(m, 0.0)
    s = m.sum()
    if s <= 0:
        return np.ones_like(m) / m.size
    return m / s

def sensitivity_entropy(m):
    p = normalize_map(m).ravel()
    eps = 1e-12
    return -np.sum(p * np.log(p + eps))

def batch_entropy(maps):
    return np.array([sensitivity_entropy(m) for m in maps], dtype=float)
```

```python
def entropy_over_checkpoints(
    ckpt_dir,
    model_fn,
    probe_images,
    device="mps",
    patch_size=4,
    stride=4,
    fill_value=0.0,
):
    ckpt_paths = sorted_checkpoint_paths(ckpt_dir)

    steps = []
    entropy_mean = []
    entropy_std = []

    for ckpt_path in ckpt_paths:
        model, step = load_checkpoint_model(ckpt_path, model_fn, device)

        maps = occlusion_maps_for_probe_set(
            model,
            probe_images,
            patch_size=patch_size,
            stride=stride,
            fill_value=fill_value,
            device=device
        )

        vals = batch_entropy(maps)

        steps.append(step)
        entropy_mean.append(vals.mean())
        entropy_std.append(vals.std())

        print(f"done step={step}, entropy={vals.mean():.4f}")

    return {
        "step": np.array(steps),
        "entropy_mean": np.array(entropy_mean),
        "entropy_std": np.array(entropy_std),
    }
```

```python
import math
import random
from dataclasses import dataclass

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt


# -----------------------------
# Reproducibility
# -----------------------------
def set_seed(seed: int = 0) -> None:
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)


# -----------------------------
# Small model where exact Jacobian is tractable
# -----------------------------
class TinyMLP(nn.Module):
    def __init__(self, input_dim: int = 16, hidden_dim: int = 32, output_dim: int = 4):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, output_dim),
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.net(x)


# -----------------------------
# True Jacobian norm
# J_f(x) in R^{k x d}
# We use Frobenius norm by default.
# You can swap to spectral norm if desired.
# -----------------------------
def true_jacobian_fro_norm(model: nn.Module, x: torch.Tensor) -> float:
    """
    Compute ||J_f(x)||_F exactly for a single input x of shape [d].
    """
    x = x.detach().clone().requires_grad_(True)

    def f(inp: torch.Tensor) -> torch.Tensor:
        return model(inp.unsqueeze(0)).squeeze(0)  # [k]

    J = torch.autograd.functional.jacobian(f, x, create_graph=False)  # [k, d]
    return torch.norm(J, p="fro").item()


def true_jacobian_spectral_norm(model: nn.Module, x: torch.Tensor) -> float:
    """
    Compute ||J_f(x)||_2 (largest singular value) exactly for a single input x.
    This is more expensive but still fine for tiny models.
    """
    x = x.detach().clone().requires_grad_(True)

    def f(inp: torch.Tensor) -> torch.Tensor:
        return model(inp.unsqueeze(0)).squeeze(0)

    J = torch.autograd.functional.jacobian(f, x, create_graph=False)  # [k, d]
    svals = torch.linalg.svdvals(J)
    return svals.max().item()


# -----------------------------
# Finite-difference sensitivity surrogate
# -----------------------------
def fd_sensitivity_single(
    model: nn.Module,
    x: torch.Tensor,
    eps: float = 1e-3,
    n_vecs: int = 8,
) -> float:
    """
    Approximate local sensitivity by random-direction finite differences:
        mean_j ||f(x + eps v_j) - f(x)|| / eps
    where ||v_j||_2 = 1.
    """
    model.eval()
    x = x.detach()

    with torch.no_grad():
        fx = model(x.unsqueeze(0)).squeeze(0)

        vals = []
        for _ in range(n_vecs):
            v = torch.randn_like(x)
            v = v / (torch.norm(v) + 1e-12)

            fx_pert = model((x + eps * v).unsqueeze(0)).squeeze(0)
            val = torch.norm(fx_pert - fx, p=2).item() / eps
            vals.append(val)

    return float(np.mean(vals))


# -----------------------------
# Sanity-check experiment
# -----------------------------
@dataclass
class SanityCheckResult:
    true_vals: np.ndarray
    fd_vals: np.ndarray
    pearson_r: float
    mse: float


def pearson_corr(x: np.ndarray, y: np.ndarray) -> float:
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    x = x - x.mean()
    y = y - y.mean()
    denom = np.sqrt((x**2).sum()) * np.sqrt((y**2).sum())
    if denom < 1e-12:
        return 0.0
    return float((x * y).sum() / denom)


def run_sanity_check(
    model: nn.Module,
    n_samples: int = 128,
    input_dim: int = 16,
    eps: float = 1e-3,
    n_vecs: int = 8,
    use_spectral: bool = False,
    device: str = "cpu",
) -> SanityCheckResult:
    model = model.to(device).eval()

    true_vals = []
    fd_vals = []

    for _ in range(n_samples):
        x = torch.randn(input_dim, device=device)

        if use_spectral:
            true_val = true_jacobian_spectral_norm(model, x)
        else:
            true_val = true_jacobian_fro_norm(model, x)

        fd_val = fd_sensitivity_single(model, x, eps=eps, n_vecs=n_vecs)

        true_vals.append(true_val)
        fd_vals.append(fd_val)

    true_vals = np.array(true_vals, dtype=float)
    fd_vals = np.array(fd_vals, dtype=float)

    r = pearson_corr(true_vals, fd_vals)
    mse = float(np.mean((true_vals - fd_vals) ** 2))

    return SanityCheckResult(
        true_vals=true_vals,
        fd_vals=fd_vals,
        pearson_r=r,
        mse=mse,
    )


# -----------------------------
# Plotting
# -----------------------------
def plot_sanity_check(result: SanityCheckResult, outfile: str = "jacobian_sanity_check.pdf") -> None:
    x = result.true_vals
    y = result.fd_vals

    lo = min(x.min(), y.min())
    hi = max(x.max(), y.max())

    plt.figure(figsize=(6.2, 5.0))
    plt.scatter(x, y, s=30, alpha=0.8)
    plt.plot([lo, hi], [lo, hi], linewidth=1.5)

    plt.xlabel("True Jacobian norm")
    plt.ylabel("Finite-difference surrogate")
    plt.title(f"Sanity check: surrogate vs. true sensitivity\nPearson r = {result.pearson_r:.3f}")

    ax = plt.gca()
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    plt.tight_layout()
    plt.savefig(outfile, dpi=300, bbox_inches="tight")
    plt.plot(sorted(x), sorted(x), '--', alpha=0.3)
    plt.show()


# -----------------------------
# Example usage
# -----------------------------
if __name__ == "__main__":
    set_seed(0)

    device = "cpu"
    model = TinyMLP(input_dim=16, hidden_dim=32, output_dim=4)

    result = run_sanity_check(
        model=model,
        n_samples=128,
        input_dim=16,
        eps=1e-2,
        n_vecs=64,
        use_spectral=False,   # set True if you want true operator norm
        device=device,
    )

    print(f"Pearson correlation: {result.pearson_r:.4f}")
    print(f"MSE: {result.mse:.6f}")

    plot_sanity_check(result, outfile="jacobian_sanity_check.pdf")
```
